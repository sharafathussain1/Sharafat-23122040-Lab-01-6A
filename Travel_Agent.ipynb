```python
print("Welcome to Smart Travel Agent")

destination = input("Where do you want to travel? ")
budget = int(input("Enter your budget (PKR): "))

packages = {
    "Dubai": 120000,
    "Turkey": 180000,
    "Hunza": 40000,
    "Murree": 25000
}

if destination in packages:
    price = packages[destination]

    print("\nDestination:", destination)
    print("Package Price: PKR", price)

    if budget >= price:
        print("You can book this package!")
        print("Have a safe and happy journey")
    else:
        print("Your budget is not enough.")
        print("Try a cheaper destination.")

else:
    print("Sorry, this destination is not available.")
```
